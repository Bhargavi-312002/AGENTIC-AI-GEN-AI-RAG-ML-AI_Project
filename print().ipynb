# Print()

```python
a=10
b=20
c=30
print() #no argument
print(a) #single argument
print(a,b,c) #multiple argument
print('the variable a is',a,'the variable b is',b,'the variable c is',c) #Result with string and variables
```

### print format method

```python
v1=10
v2=20
print('The v1 {} and v2 {} are two variables'.format(v1,v2))
```

```python
v1=10
v2=20
print('The v1 {} and v2 {} are two variables'.format(v1)) #gets an error because there are 2 {} but only one variable is mentioned
```

### f'string method

```python
v1=10
v2=10
print(f'The v1 {v1} and v2 {v2} are two variables')
```

```python
print("hello i'am")
print(""" hello I'am bhargavi """) # """ is used when ' required in the output
print(""" hello I"am bhargavi """) # """ is used when " required in the output
print(" hello I\'\ am \"bhargavi\" ") # by mentioning ' in \\ we can print ' in output
print(" hello I\"am bhargavi ") # by mentioning " in \ we can print " in output
print(""" hello I"am "bhargavi" """)
```

### End statement

End statement is used to stop generating next line. 

syntax: print('text',end='')

it will merge the lines by using the text mentioned in the end statement

```python
print('hello')
print("work on it")
print('hello',end='') #merge lines without space as nothing is mentioned in the end statement
print("work on it")
print('hello',end=' ')
print("work on it")
print('hello',end='*') #merge line with *
print("work on it")
```

```python
print(""" \"" hello \"""bhargavi\""" \"" """)
```

```python
print(" \"\"\"\" hello bhargavi\"\"\"\" ")
```

```python
print('hello',end=4) 
print("work on it")
```
