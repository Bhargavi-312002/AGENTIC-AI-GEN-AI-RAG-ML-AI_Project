##### Index slicing

##### Forward slicing

```python
print(si[:]) #print complete string because start and end index are not mentioned
print(si[1:4]) #prints the character from index 1 to index 3. 
print(si[3:10]) # we will not get error because the end index is max of length of the string so it print the characters from index 3 to last 
print(si[2:]) # prints the string from index 2 to end
print(si[3:]) #prints the string from index 3 to end
print(si[:6]) #prints the index from starting to index 5
print(si[:10])
```
