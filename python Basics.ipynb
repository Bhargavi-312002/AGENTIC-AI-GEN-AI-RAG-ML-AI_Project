##### Index slicing

##### Forward slicing

```python
print(si[:]) #print complete string because start and end index are not mentioned
print(si[1:4]) #prints the character from index 1 to index 3. 
print(si[3:10]) # we will not get error because the end index is max of length of the string so it print the characters from index 3 to last 
print(si[2:]) # prints the string from index 2 to end
print(si[3:]) #prints the string from index 3 to end
print(si[:6]) #prints the index from starting to index 5
print(si[:10])
```

##### Step slicing

```python
si='welcome'
print(si[0:7:2]) #It starts printing from index 0 to 6 with the step size of 2 i.e., first it prints w as step is 2 skip e and print the l.
print(si[0:20:4]) #it prints from string with index 4
print(si[::2]) #prints from start to end with index 2 as start index and end index not mentioned
print(si[::3])
print(si[::-1]) #it reverse the string as the step is -1
print(si[::-2]) #it prints from last with step -2
```

##### Backward slicing

```python
si='welcome'
si[-5:-1] #starts printing from -5 index to -2 index
```

## Python Type casting

#### 1. int(value):

```python
a=int(10.20) #converts float to int
b=int(True)  #converts Boolean to int
c=int(False) #converts Boolean to int
d=int('10')  # converts number string to int
print((a))
print((b))
print((c))
print((d))
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
print(int(1+2j))
```

```python
print(int('ten'))
```

#### 2. Float(Value):

```python
a=float(10) #converts int to float
b=float(True)  #converts Boolean to float
c=float(False) #converts Boolean to float
d=float('10')  # converts number string to float
print((a))
print((b))
print((c))
print((d))
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
print(float(10+2j))
```

```python
print(float('ten'))
```

#### 3. bool(value)

```python
# Except 0 bool will consider all others as True

a=bool(10) #converts int to Boolean
b=bool(0) #converts int to Boolean
c=bool(10.2)  #converts float to boolen
d=bool(0.0)  #converts float to boolen
e=bool(1+2j)  #converts complex to boolen
f=bool(0+0j)  #converts complex to boolen
g=bool('ten') #converts string to boolen
h=bool('10') #converts string to boolen
i=bool( ) #we will output as False for emptyspace
print('10 boolean is:',a)
print(('bool(0) is :',b))
print(('bool(10.2) is :',c))
print(('bool(0.0) is :',d))
print(('bool(1+2j) is :',e))
print(('bool(0+0j) is :',f))
print('bool(ten) is:',g)
print('bool(string(10) is:',h)
print('for empty space:',i)
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
print(type(i))
```

#### 3. Complex(value):

```python
a=complex(10) #converts int to complex
b=complex(2,3) #converts int to complex
c=complex(10.2)  #converts float to complex
d=complex(10.0,5.3)  #converts float to complex
e=complex(True)  #converts boolean to complex
f=complex(False)  #converts boolean to complex
g=complex('10') #converts string to complex
print('complex(10) is:',a)
print(('complex(2,3) is :',b))
print(('complex(10.2) is :',c))
print(('complex(0.0,5.3) is :',d))
print(('complex(True) is :',e))
print(('complex(False) is :',f))
print('complex(string(10) is:',g)
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
```

```python
print(complex('ten')) #number string cannot be converted to complex
```

```python
complex(3,4,5) #3 arguments cannot be accepted
```
